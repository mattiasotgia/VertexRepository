# Monitor Infrastructure usage

This notebooks aims at analysing the resource usage for the DL vertex in Pandora for ICARUS. 

This will be key for defining a standard usage for the `Stage1` processing.

```python
import sqlite3 as sql
import pandas as pd
```

```python
# Connect to the database
# conn = sql.connect("../data/MemoryReport.db")
conn = sql.connect("../data/TimingReport.db")

# Get the names of all tables
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';",
    conn
)["name"].tolist()

# Load each table into a DataFrame, stored in a dict keyed by table name
dfs = {name: pd.read_sql_query(f'SELECT * FROM "{name}"', conn) for name in tables}

conn.close()

# Usage
print(list(dfs.keys()))        # table names
dfs[tables[2]]['Event == 17']
```
